### STEP 1: Import Pandas as pd

```python
import pandas as pd
```

### STEP 2: import data from github

```python
url = "https://raw.githubusercontent.com/ritaafrica/data/refs/heads/main/student_scores.csv"
df = pd.read_csv(url)
df.head()
```

### STEP 3: Add new column called "Total scores" to the dataset

```python
df ["Status"] = "Pending" # demo
```

```python
df.head()
```

```python
df ["Total_Scores"] = df["Math_Score"] + df["Reading_Score"] + df["Writing_Score"]
```

```python
df.head()
```

### STEP 4: Calculate the percentage_scores

```python
df ["percentage_scores"] = df ["Total Scores"] /300*100
df ["percentage_scores"] = round(df ["Total Scores"] /300*100) # to roundoff
```

```python
df.head()
```

### STEP 5: Rename Math_Score to Mathematics_Score

```python
df.rename(columns={"Math_Score" : "Mathematics_Score"}, inplace=True)
```

```python
df.head()
```

### STEP 6: Rename Gender to G

```python
df.rename(columns={"Gender" : "G"}, inplace=True)
df.head()
```

### STEP 7: Remove/ delete the "School" column 

```python
df.drop(columns=["School"], inplace=True)
df.head()
```

### STEP 8: save the dataset to csv

```python
df.to_csv("modified_student_scores.csv", index=False)
```
